Создать базовый класс Book в C#, который будет представлять информацию о
книгах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

### Требования к базовому классу Book:
• Атрибуты: Название (Title), Автор (Author), Год издания (YearOfPublication).

• Методы:

    • GetInfo(): метод для получения информации о книге в виде строки.
    • Read(): метод для вывода сообщения о чтении книги.
    • Borrow(): метод для вывода сообщения о выдаче книги на чтение.

### Требования к производным классам:
1. Учебник (Textbook): Должен содержать дополнительные атрибуты, такие как
Предмет (Subject). Метод Read() должен быть переопределен для
добавления информации о предмете при чтении учебника.
2. Художественная литература (Fiction): Должен содержать дополнительные
атрибуты, такие как Жанр (Genre). Метод Borrow() должен быть
переопределен для добавления информации о жанре при выдаче книги на
чтение.
3. Научная литература (ScientificLiterature) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Область науки
(FieldOfScience). Метод GetInfo() должен быть переопределен для включения
информации об области науки в описании книги.

```
using System;
using System.Collections.Generic;
using System.Linq;

public interface ICopiable
{
    void MakeCopy();
    int CopyCount { get; set; }
}

public interface IAnnotatable
{
    void AddAnnotation(string note);
    string Annotation { get; set; }
}

public interface IBookStorage
{
    void Add(Book book);
    bool Remove(Book book);
    bool Contains(Book book);
}

public delegate void BookHandler(Book book);
public delegate void BookPriceHandler(Book book, decimal oldPrice, decimal newPrice);
public delegate void MessageHandler(string message);

public class Book : ICopiable
{
    public string Title { get; set; }
    public string Author { get; set; }
    public int YearOfPublication { get; set; }
    public int PageCount { get; set; }
    public decimal Price { get; set; }
    public bool IsAvailable { get; set; } = true;
    public int CopyCount { get; set; } = 1;
    public string Isbn { get; set; } = "978-0-00-000000-0";
    public string Publisher { get; set; } = "Неизвестно";
    public string Language { get; set; } = "Русский";
    public int ReaderAgeLimit { get; set; } = 0;
    public DateTime AddedToLibrary { get; set; } = DateTime.Now;
    public int TimesRead { get; set; } = 0;
    public string ShelfLocation { get; set; } = "A-01";
    public bool IsReserved { get; set; } = false;

    public event BookHandler BookOpened;
    public event BookHandler BookArchived;
    public event BookPriceHandler PriceChanged;
    public event MessageHandler StatusChanged;

    public Book(string title, string author, int year, int pages, decimal price)
    {
        Title = title;
        Author = author;
        YearOfPublication = year;
        PageCount = pages;
        Price = price;
    }

    public void Open()
    {
        TimesRead++;
        Console.WriteLine($"Книга '{Title}' открыта.");
        BookOpened?.Invoke(this);
    }
    public void Open(int page) => Console.WriteLine($"Книга '{Title}' открыта на странице {page}.");
    public void Close() => Console.WriteLine($"Книга '{Title}' закрыта.");
    public void UpdatePrice(decimal newPrice)
    {
        var old = Price;
        Price = newPrice;
        Console.WriteLine($"Цена '{Title}' обновлена до {Price:N2} руб.");
        PriceChanged?.Invoke(this, old, newPrice);
    }
    public void Archive()
    {
        Console.WriteLine($"Книга '{Title}' отправлена в архив.");
        BookArchived?.Invoke(this);
    }

    public void ShowIsbn() => Console.WriteLine($"ISBN '{Title}': {Isbn}");
    public void ChangePublisher(string publisher) { Publisher = publisher; Console.WriteLine($"Издатель '{Title}' изменён на {Publisher}."); }
    public void ShowLanguage() => Console.WriteLine($"Язык книги '{Title}': {Language}");
    public void ShowAgeLimit() => Console.WriteLine($"Возрастное ограничение '{Title}': {ReaderAgeLimit}+");

    public void Reserve() { IsReserved = true; StatusChanged?.Invoke($"Книга '{Title}' зарезервирована."); }
    public void Unreserve() { IsReserved = false; StatusChanged?.Invoke($"Резерв с '{Title}' снят."); }
    public void MoveToShelf(string shelf) { ShelfLocation = shelf; StatusChanged?.Invoke($"'{Title}' перемещена на полку {shelf}."); }
    public void ShowStatistics() => Console.WriteLine($"'{Title}': прочитана {TimesRead} раз, добавлена {AddedToLibrary:dd.MM.yyyy}, полка {ShelfLocation}.");

    void ICopiable.MakeCopy()
    {
        CopyCount++;
        Console.WriteLine($"[ICopiable] Создана копия книги '{Title}'. Всего копий: {CopyCount}.");
    }

    int ICopiable.CopyCount
    {
        get => CopyCount;
        set => CopyCount = value;
    }

    public virtual void MakeCopy() => ((ICopiable)this).MakeCopy();

    public virtual string GetInfo() =>
        $"Книга: '{Title}', Автор: {Author}, Год: {YearOfPublication}, Страниц: {PageCount}, Цена: {Price:N2} руб.";

    public virtual void Read() => Console.WriteLine($"Читаем книгу '{Title}'...");
    public virtual void Read(int pages) => Console.WriteLine($"Читаем '{Title}' — {pages} страниц.");
    public virtual void Read(string note) => Console.WriteLine($"Читаем '{Title}' ({note}).");

    public virtual void Borrow() => Console.WriteLine($"Книга '{Title}' выдана на чтение.");
}

public class Textbook : Book
{
    public string Subject { get; set; }
    public string EducationLevel { get; set; }
    public bool HasPracticalTasks { get; set; }
    public string ApprovalCode { get; set; }
    public int ExercisesCount { get; set; }
    public string Ministry { get; set; } = "Минпросвещения РФ";
    public int Edition { get; set; } = 1;
    public bool HasElectronicVersion { get; set; } = false;
    public string Udk { get; set; } = "373";
    public string Curriculum { get; set; } = "ФГОС";
    public int HoursPerWeek { get; set; } = 2;
    public bool HasAnswerKey { get; set; } = true;
    public string CoverType { get; set; } = "Твёрдый";

    public Textbook(string title, string author, int year, int pages, decimal price,
                    string subject, string level, bool practical,
                    string approval = "FGOS-2024", int exercises = 100)
        : base(title, author, year, pages, price)
    {
        Subject = subject;
        EducationLevel = level;
        HasPracticalTasks = practical;
        ApprovalCode = approval;
        ExercisesCount = exercises;
    }

    public void CheckCurriculum() => Console.WriteLine($"Проверка '{Title}' на соответствие ФГОС по предмету {Subject}.");
    public void SolveTest() => Console.WriteLine($"Решение упражнений из учебника '{Title}'.");
    public void AssignHomework(int chapter) => Console.WriteLine($"Задано ДЗ по главе {chapter} из '{Title}'.");
    public void VerifyApproval() => Console.WriteLine($"Код '{ApprovalCode}' для '{Title}' действителен.");

    public void ShowMinistry() => Console.WriteLine($"Учебник '{Title}' рекомендован: {Ministry}.");
    public void IncrementEdition() { Edition++; Console.WriteLine($"Издание '{Title}': {Edition}-е."); }
    public void EnableElectronicVersion() { HasElectronicVersion = true; Console.WriteLine($"'{Title}' имеет электронную версию."); }
    public void ShowUdk() => Console.WriteLine($"УДК '{Title}': {Udk}");

    public void ShowCurriculum() => Console.WriteLine($"Программа '{Title}': {Curriculum}, {HoursPerWeek} ч/нед.");
    public void ToggleAnswerKey() { HasAnswerKey = !HasAnswerKey; Console.WriteLine($"Ключи ответов: {(HasAnswerKey ? "есть" : "нет")}."); }
    public void ChangeCover(string cover) { CoverType = cover; Console.WriteLine($"Обложка '{Title}': {CoverType}."); }
    public void ShowWeeklyLoad() => Console.WriteLine($"'{Title}' — {HoursPerWeek} часов в неделю.");

    public override string GetInfo() =>
        $"{base.GetInfo()}, Предмет: {Subject}, Уровень: {EducationLevel}";

    public override void Read()
    {
        base.Read();
        Console.WriteLine($"  → учебник по '{Subject}' для {EducationLevel}.");
    }
}

public class DigitalTextbook : Textbook
{
    public string FileFormat { get; set; }
    public double FileSizeMB { get; set; }
    public string CloudUrl { get; set; }
    public string EncryptionType { get; set; }
    public int DownloadCount { get; set; }
    public bool RequiresVpn { get; set; }
    public bool IsDrmProtected { get; set; } = true;
    public string Platform { get; set; } = "LMS Moodle";
    public int LicenseDays { get; set; } = 365;
    public string Checksum { get; set; } = "sha256:...";
    public string AccessCode { get; set; } = "EDU-0000";
    public bool OfflineMode { get; set; } = false;
    public int MaxDevices { get; set; } = 3;
    public DateTime LastSync { get; set; } = DateTime.Now;

    public DigitalTextbook(string title, string author, int year, int pages, decimal price,
                           string subject, string level, bool practical,
                           string format, double size,
                           string url = "cloud.edu", string encryption = "AES-256")
        : base(title, author, year, pages, price, subject, level, practical)
    {
        FileFormat = format;
        FileSizeMB = size;
        CloudUrl = url;
        EncryptionType = encryption;
        RequiresVpn = true;
    }

    public void Download() => Console.WriteLine($"Загрузка '{Title}' в формате {FileFormat} ({FileSizeMB} MB)...");
    public void SyncWithCloud() => Console.WriteLine($"Синхронизация '{Title}' с {CloudUrl}...");
    public void Decrypt() => Console.WriteLine($"Расшифровка файла ({EncryptionType})...");
    public void IncrementDownloads() { DownloadCount++; Console.WriteLine($"Скачиваний '{Title}': {DownloadCount}."); }

    public void ShowDrm() => Console.WriteLine($"DRM '{Title}': {(IsDrmProtected ? "включена" : "отключена")}.");
    public void ChangePlatform(string platform) { Platform = platform; Console.WriteLine($"Платформа '{Title}': {Platform}."); }
    public void ExtendLicense(int days) { LicenseDays += days; Console.WriteLine($"Лицензия '{Title}' продлена на {days} дн. (итого {LicenseDays})."); }
    public bool VerifyChecksum() { Console.WriteLine($"Проверка контрольной суммы '{Title}'..."); return true; }

    public void EnableOffline() { OfflineMode = true; Console.WriteLine($"'{Title}' доступен офлайн."); }
    public void ShowAccessCode() => Console.WriteLine($"Код доступа '{Title}': {AccessCode}.");
    public void LimitDevices(int n) { MaxDevices = n; Console.WriteLine($"Устройств для '{Title}': {MaxDevices}."); }
    public void ShowLastSync() => Console.WriteLine($"Последняя синхронизация '{Title}': {LastSync:g}.");

    public override string GetInfo() =>
        $"{base.GetInfo()}, Формат: {FileFormat}, Размер: {FileSizeMB} MB";

    public void Download(bool highQuality) =>
        Console.WriteLine(highQuality
            ? $"Загрузка '{Title}' в HD-качестве ({FileSizeMB * 1.5} MB)..."
            : $"Загрузка '{Title}' в стандартном качестве ({FileSizeMB} MB)...");
}

public class Fiction : Book, IAnnotatable
{
    public string Genre { get; set; }
    public string AgeRating { get; set; }
    public bool IsBestseller { get; set; }
    public string OriginalLanguage { get; set; }
    public int AwardsCount { get; set; }
    public string Annotation { get; set; } = "";
    public string Setting { get; set; } = "Россия";
    public int ChaptersCount { get; set; } = 20;
    public string MainCharacter { get; set; } = "Неизвестен";
    public bool IsAdaptedToFilm { get; set; } = false;
    public string NarrativeStyle { get; set; } = "Реализм";
    public int PagesRead { get; set; } = 0;
    public bool IsSeries { get; set; } = false;
    public string SeriesName { get; set; } = "";

    public Fiction(string title, string author, int year, int pages, decimal price,
                   string genre, string ageRating, bool bestseller,
                   string language = "Русский", int awards = 0)
        : base(title, author, year, pages, price)
    {
        Genre = genre;
        AgeRating = ageRating;
        IsBestseller = bestseller;
        OriginalLanguage = language;
        AwardsCount = awards;
    }

    public void AnalyzePlot() => Console.WriteLine($"Анализ сюжета книги '{Title}'.");
    public void ReviewAgeRating() => Console.WriteLine($"Возрастной рейтинг '{Title}': {AgeRating}.");
    public void MarkBestseller() { IsBestseller = true; Console.WriteLine($"'{Title}' отмечена как бестселлер."); }
    public void TranslateTo(string lang) => Console.WriteLine($"'{Title}' переведена на {lang}.");

    public void ShowSetting() => Console.WriteLine($"Место действия '{Title}': {Setting}.");
    public void AddChapter() { ChaptersCount++; Console.WriteLine($"Глав в '{Title}': {ChaptersCount}."); }
    public void SetMainCharacter(string name) { MainCharacter = name; Console.WriteLine($"Главный герой '{Title}': {MainCharacter}."); }
    public void AdaptToFilm() { IsAdaptedToFilm = true; Console.WriteLine($"'{Title}' экранизирована."); }

    public void ShowNarrativeStyle() => Console.WriteLine($"Стиль '{Title}': {NarrativeStyle}.");
    public void ReadPages(int n) { PagesRead += n; Console.WriteLine($"Прочитано '{Title}': {PagesRead} стр."); }
    public void MarkAsSeries(string name) { IsSeries = true; SeriesName = name; Console.WriteLine($"'{Title}' — часть серии '{name}'."); }
    public void ShowReadingProgress() => Console.WriteLine($"Прогресс '{Title}': {PagesRead}/{PageCount} стр.");

    void IAnnotatable.AddAnnotation(string note)
    {
        Annotation = note;
        Console.WriteLine($"Аннотация к '{Title}': \"{note}\"");
    }

    string IAnnotatable.Annotation
    {
        get => Annotation;
        set => Annotation = value;
    }

    public void AddAnnotation(string note) => ((IAnnotatable)this).AddAnnotation(note);
    public void AddAnnotation() => AddAnnotation("Без комментария.");

    public override string GetInfo() =>
        $"{base.GetInfo()}, Жанр: {Genre}, Возраст: {AgeRating}, Бестселлер: {(IsBestseller ? "Да" : "Нет")}";

    public override void Borrow() =>
        Console.WriteLine($"Выдана художественная книга (Жанр: {Genre}, Возраст: {AgeRating}) — '{Title}'.");
}

public class ScientificLiterature : Book, IAnnotatable
{
    public string FieldOfScience { get; set; }
    public string ResearchInstitution { get; set; }
    public bool PeerReviewed { get; set; }
    public string Doi { get; set; }
    public int ReferenceCount { get; set; }
    public string Annotation { get; set; } = "";
    public string Issn { get; set; } = "0000-0000";
    public int ImpactFactor { get; set; } = 3;
    public string ReviewBoard { get; set; } = "Редакционная коллегия";
    public bool IsOpenAccess { get; set; } = true;
    public string CitationStyle { get; set; } = "APA";
    public int HIndex { get; set; } = 0;
    public bool IsIndexedScopus { get; set; } = false;
    public string ConferenceName { get; set; } = "";

    public ScientificLiterature(string title, string author, int year, int pages, decimal price,
                                string field, string institution, bool peerReviewed,
                                string doi = "10.1000/xyz", int references = 50)
        : base(title, author, year, pages, price)
    {
        FieldOfScience = field;
        ResearchInstitution = institution;
        PeerReviewed = peerReviewed;
        Doi = doi;
        ReferenceCount = references;
    }

    public void VerifyCitations() => Console.WriteLine($"Проверка цитирований в '{Title}'.");
    public void PresentResearch() => Console.WriteLine($"Презентация исследования '{Title}' ({FieldOfScience}).");
    public void ShowDoi() => Console.WriteLine($"DOI: {Doi}.");
    public void ListReferences() => Console.WriteLine($"Источников: {ReferenceCount}.");

    public void ShowIssn() => Console.WriteLine($"ISSN '{Title}': {Issn}.");
    public void UpdateImpactFactor(int value) { ImpactFactor = value; Console.WriteLine($"Impact Factor '{Title}': {ImpactFactor}."); }
    public void ShowReviewBoard() => Console.WriteLine($"Рецензенты '{Title}': {ReviewBoard}.");
    public void EnableOpenAccess() { IsOpenAccess = true; Console.WriteLine($"'{Title}' в открытом доступе."); }

    public void ShowCitationStyle() => Console.WriteLine($"Стиль цитирования '{Title}': {CitationStyle}.");
    public void UpdateHIndex(int h) { HIndex = h; Console.WriteLine($"h-index '{Title}': {HIndex}."); }
    public void IndexInScopus() { IsIndexedScopus = true; Console.WriteLine($"'{Title}' проиндексирована в Scopus."); }
    public void SetConference(string name) { ConferenceName = name; Console.WriteLine($"'{Title}' представлена на '{name}'."); }

    void IAnnotatable.AddAnnotation(string note)
    {
        Annotation = note;
        Console.WriteLine($"Научная рецензия к '{Title}': \"{note}\"");
    }

    string IAnnotatable.Annotation
    {
        get => Annotation;
        set => Annotation = value;
    }

    public void AddAnnotation(string note) => ((IAnnotatable)this).AddAnnotation(note);
    public void AddAnnotation(int rating) => AddAnnotation($"Оценка рецензента: {rating}/10.");

    public override string GetInfo() =>
        $"{base.GetInfo()}, Область: {FieldOfScience}, Институт: {ResearchInstitution}, Рецензия: {(PeerReviewed ? "Да" : "Нет")}";
}

public class Catalog<T> where T : Book
{
    private List<T> items = new List<T>();
    private Dictionary<string, T> byTitle = new Dictionary<string, T>();
    private HashSet<string> authors = new HashSet<string>();

    public event Action<T> ItemAdded;

    public void Add(T item)
    {
        items.Add(item);
        byTitle[item.Title] = item;
        authors.Add(item.Author);
        ItemAdded?.Invoke(item);
    }

    public void ShowAll()
    {
        Console.WriteLine($"--- Каталог ({typeof(T).Name}, {items.Count} шт.) ---");
        foreach (var item in items) Console.WriteLine(item.GetInfo());
    }

    public T FindByTitle(string title) => byTitle.TryGetValue(title, out var b) ? b : null;

    public void ShowAuthors()
    {
        Console.WriteLine("Авторы в каталоге:");
        foreach (var a in authors) Console.WriteLine($" - {a}");
    }

    public int Count => items.Count;
}

public class Repository<T> where T : Book
{
    private Dictionary<string, T> storage = new Dictionary<string, T>();

    public void Save(T item) => storage[item.Title] = item;

    public T Get(string title)
    {
        storage.TryGetValue(title, out var found);
        return found;
    }

    public void ShowAllTitles()
    {
        Console.WriteLine($"=== Хранилище ({typeof(T).Name}) ===");
        foreach (var key in storage.Keys) Console.WriteLine($" - {key}");
    }
}

public class InMemoryBookStorage : IBookStorage
{
    private readonly List<Book> books = new List<Book>();
    public void Add(Book book) => books.Add(book);
    public bool Remove(Book book) => books.Remove(book);
    public bool Contains(Book book) => books.Contains(book);
}

public class Reader
{
    public string Name { get; set; }
    private List<Book> borrowedBooks = new List<Book>();

    public Reader(string name) => Name = name;

    public void BorrowBook(Book book)
    {
        borrowedBooks.Add(book);
        Console.WriteLine($"{Name} взял книгу: '{book.Title}'");
        book.Borrow();
    }

    public void ReadBook(Book book)
    {
        Console.WriteLine($"{Name} читает...");
        book.Read();
    }

    public void ReadBook(Book book, int pages)
    {
        Console.WriteLine($"{Name} читает {pages} стр. из книги...");
        book.Read(pages);
    }

    public void ShowBorrowed()
    {
        Console.WriteLine($"У {Name} на руках: {borrowedBooks.Count} книг.");
        foreach (var b in borrowedBooks) Console.WriteLine($"  - {b.Title}");
    }
}

public class Library
{
    public string Name { get; set; }
    private readonly IBookStorage _storage;

    public Library(string name, IBookStorage storage)
    {
        Name = name;
        _storage = storage;
    }

    public void AddBook(Book book)
    {
        _storage.Add(book);
        Console.WriteLine($"В библиотеку '{Name}' добавлена книга: '{book.Title}'");
    }

    public void LendBook(Book book, Reader reader)
    {
        if (!_storage.Contains(book))
        {
            Console.WriteLine($"Книги '{book.Title}' нет в библиотеке!");
            return;
        }
        _storage.Remove(book);
        Console.WriteLine($"Библиотека '{Name}' выдает '{book.Title}' читателю {reader.Name}");
        reader.BorrowBook(book);
    }

    public void LendBook(Book book) => LendBook(book, new Reader("Гость"));
}

public delegate void DisplayDelegate();
public delegate void AnimalHandler(string message);

public class Animal
{
    public string Name { get; set; }
    public string Poroda { get; set; }
    public int Age { get; set; } = 1;
    public double Weight { get; set; } = 5.0;
    public string Color { get; set; } = "Неизвестно";
    public bool IsVaccinated { get; set; } = false;

    public Animal(string name, string poroda)
    {
        Name = name;
        Poroda = poroda;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Порода: {Poroda}");
    }

    public event AnimalHandler AnimalRename;
    public event AnimalHandler AgeChanged;

    public void Renamed(string newName)
    {
        Name = newName;
        AnimalRename?.Invoke($"Имя изменено на {newName}");
    }

    public void SetAge(int age)
    {
        int old = Age;
        Age = age;
        AgeChanged?.Invoke($"[{Name}] Возраст: {old} → {age}");
    }
    public void Vaccinate() { IsVaccinated = true; Console.WriteLine($"{Name} вакцинирован."); }
    public void SetColor(string color) { Color = color; Console.WriteLine($"{Name}: окрас {color}."); }
    public void ShowWeight() => Console.WriteLine($"{Name}: вес {Weight} кг, возраст {Age} г.");
}

public class Cat : Animal
{
    public int CountPaws { get; set; }
    public bool IsIndoor { get; set; } = true;
    public string FavoriteFood { get; set; } = "Рыба";
    public int LivesLeft { get; set; } = 9;
    public bool CanHunt { get; set; } = true;

    public Cat(string name, string poroda, int cntpaws)
        : base(name, poroda)
    {
        CountPaws = cntpaws;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество лап: {CountPaws}");
    }

    public void SetFavoriteFood(string f) { FavoriteFood = f; Console.WriteLine($"Любимая еда {Name}: {f}."); }
    public void LoseLife() { LivesLeft--; Console.WriteLine($"{Name}: осталось жизней {LivesLeft}."); }
    public void ToggleIndoor() { IsIndoor = !IsIndoor; Console.WriteLine($"{Name}: {(IsIndoor ? "домашняя" : "уличная")}."); }
    public void ShowHunting() => Console.WriteLine($"{Name} {(CanHunt ? "умеет" : "не умеет")} охотиться.");
}

public class Dog : Animal
{
    public bool HasHome { get; set; }
    public string Breed { get; set; } = "";
    public bool IsTrained { get; set; } = false;
    public int WalkCount { get; set; } = 0;
    public string FavoriteCommand { get; set; } = "Сидеть";

    public Dog(string model, string manufacturer, bool hashome)
        : base(model, manufacturer)
    {
        HasHome = hashome;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет дом: {HasHome}");
    }

    public void Train() { IsTrained = true; Console.WriteLine($"{Name} обучен команде '{FavoriteCommand}'."); }
    public void Walk() { WalkCount++; Console.WriteLine($"{Name} выгулян. Всего прогулок: {WalkCount}."); }
    public void SetCommand(string cmd) { FavoriteCommand = cmd; Console.WriteLine($"Новая команда для {Name}: {cmd}."); }
    public void ShowTraining() => Console.WriteLine($"{Name}: обучен = {IsTrained}.");
}

class Program
{
    static void Main()
    {
        var library = new Book[]
        {
            new Textbook("Алгебра 9 класс", "А.Г. Мордкович", 2020, 250, 750.0m, "Математика", "Средняя школа", true),
            new DigitalTextbook("Физика 10 класс", "М.Я. Кандауров", 2023, 300, 450.0m, "Физика", "Старшая школа", true, "PDF", 15.4),
            new Fiction("Мастер и Маргарита", "М.А. Булгаков", 1967, 480, 990.0m, "Роман", "16+", true),
            new ScientificLiterature("Краткая история времени", "Стивен Хокинг", 1988, 220, 1200.0m, "Астрофизика", "Cambridge University", true)
        };

        var demos = new (string Title, Action Body)[]
        {
            ("ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (перекрытие)", () =>
            {
                foreach (var book in library)
                {
                    Console.WriteLine(book.GetInfo());
                    book.Read();
                    book.Borrow();
                    book.MakeCopy();
                    book.Read(10);
                    book.Read("вдумчиво");
                    if (book is IAnnotatable annotatable)
                        annotatable.AddAnnotation("Обязательно к прочтению.");
                    Console.WriteLine("--------------------------------------------------");
                }
            }),

            ("ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ", () =>
            {
                var b = library[1];
                b.Open();
                b.Open(42);
                ((DigitalTextbook)b).Download();
                ((DigitalTextbook)b).Download(true);
            }),

            ("НОВЫЕ АТРИБУТЫ И МЕТОДЫ", () =>
            {
                var b = library[1];
                b.ShowIsbn();
                b.ChangePublisher("Просвещение");
                b.ShowLanguage();
                b.ShowAgeLimit();
                ((Textbook)library[0]).ShowMinistry();
                ((DigitalTextbook)library[1]).ShowDrm();
                ((Fiction)library[2]).ShowSetting();
                ((ScientificLiterature)library[3]).ShowIssn();
            }),

            ("ДЕЛЕГАТЫ И СОБЫТИЯ", () =>
            {
                var b = library[0];
                BookHandler onOpen = x => Console.WriteLine($"[Подписка] Открыта: {x.Title} (раз: {x.TimesRead})");
                b.BookOpened += onOpen;
                b.PriceChanged += (x, o, n) => Console.WriteLine($"[Цена] '{x.Title}': {o:N2} → {n:N2} руб.");
                b.StatusChanged += m => Console.WriteLine($"[Статус] {m}");
                b.Open();
                b.UpdatePrice(850m);
                b.Reserve();
                b.MoveToShelf("B-07");
                b.ShowStatistics();
            }),

            ("НОВЫЕ МЕТОДЫ (ПРОИЗВОДНЫЕ)", () =>
            {
                ((Textbook)library[0]).ShowCurriculum();
                ((Fiction)library[2]).ReadPages(75);
                ((Fiction)library[2]).ShowReadingProgress();
                ((ScientificLiterature)library[3]).UpdateHIndex(45);
            }),

            ("GENERIC КАТАЛОГ", () =>
            {
                var catalog = new Catalog<Book>();
                catalog.ItemAdded += b => Console.WriteLine($"[Catalog] + {b.Title}");
                foreach (var b in library) catalog.Add(b);
                catalog.ShowAll();
                catalog.ShowAuthors();
                Console.WriteLine($"Найдено: {catalog.FindByTitle("Мастер и Маргарита")?.GetInfo()}");

                var fc = new Catalog<Fiction>();
                fc.Add((Fiction)library[2]);
                fc.ShowAll();
            }),

            ("GENERIC РЕПОЗИТОРИЙ", () =>
            {
                var repo = new Repository<Book>();
                foreach (var b in library) repo.Save(b);
                repo.ShowAllTitles();
                Console.WriteLine($"Из хранилища: {repo.Get("Краткая история времени").GetInfo()}");
            }),

            ("ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ (С УПРАВЛЕНИЕМ ЗАВИСИМОСТЯМИ)", () =>
            {
                IBookStorage storage = new InMemoryBookStorage();
                var lib = new Library("Городская библиотека", storage);
                foreach (var b in library) lib.AddBook(b);

                var reader = new Reader("Иван");
                lib.LendBook(library[2], reader);
                reader.ReadBook(library[2]);
                reader.ReadBook(library[2], 50);
                reader.ShowBorrowed();
                lib.LendBook(library[0]);
            }),

            ("ЖИВОТНЫЕ: ДЕЛЕГАТЫ, СОБЫТИЯ, КОЛЛЕКЦИИ", () =>
            {
                AnimalHandler logger = m => Console.WriteLine($"[LOG] {m}");

                var test = new Animal("Бобик", "Дворняга");
                test.AnimalRename += logger;
                test.AgeChanged += logger;
                test.SetAge(3);
                test.Renamed("Барсик");
                test.ShowWeight();

                List<Animal> zoo = new()
                {
                    new Cat("Мурка", "Персидская", 4) { Weight = 4.2, Color = "Белый" },
                    new Dog("Шарик", "Овчарка", true) { Weight = 25.0, Color = "Коричневый" },
                    new Cat("Том", "Британская", 4) { Weight = 5.1, Color = "Серый" }
                };

                var byName = new Dictionary<string, Animal>();
                foreach (var a in zoo)
                {
                    byName[a.Name] = a;
                    a.AgeChanged += logger;
                    a.AnimalRename += logger;
                }
                Console.WriteLine($"Животных в зоопарке: {byName.Count}");

                Console.WriteLine("\n--- Только домашние коты (лямбда + LINQ) ---");
                foreach (var c in zoo.OfType<Cat>().Where(c => c.IsIndoor))
                    c.DisplayInfo();

                Console.WriteLine("\n--- Изменения через события ---");
                var murka = (Cat)byName["Мурка"];
                murka.SetAge(4);
                murka.SetFavoriteFood("Курица");
                murka.Renamed("Муся");

                var sharik = (Dog)byName["Шарик"];
                sharik.Walk();
                sharik.Train();
                sharik.SetCommand("Лежать");

                Console.WriteLine("\n--- Группировка ---");
                foreach (var g in zoo.GroupBy(a => a.GetType().Name))
                    Console.WriteLine($"{g.Key}: {g.Count()} шт. — {string.Join(", ", g.Select(x => x.Name))}");

                Console.WriteLine("\n--- Делегаты ---");
                DisplayDelegate dc = new Cat("Муся", "Сиамская", 4).DisplayInfo;
                DisplayDelegate dd = new Dog("Рэкс", "Питбуль", true).DisplayInfo;
                foreach (var d in new[] { dc, dd }) d();
            })
        };

        foreach (var (title, body) in demos)
        {
            Console.WriteLine($"\n=== {title} ===");
            body();
        }
    }
}
```
