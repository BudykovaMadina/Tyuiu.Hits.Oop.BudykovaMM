Дан целочисленный массив numbers со следующими элементами { 5, 8, 1, 3, 9, 6, 7, 2 }, нужно выполнить следующие операции:
1.	Найти минимальное и максимальное значения в массиве.
2.	Найти сумму всех элементов массива.
3.	Найти среднее арифметическое всех элементов массива.
4.	Отсортировать массив в порядке возрастания.
5.	Проверить, содержит ли массив определенное число.


```polyglot-notebook
int[] numbers = { 5, 8, 1, 3, 9, 6, 7, 2 };
```

```polyglot-notebook
using System.Linq;

int max = numbers.Max();
int min = numbers.Min();

Console.WriteLine($"Минимальное значение в массиве: {min}");
Console.WriteLine($"Максимальное значение в массиве: {max}");
```

```polyglot-notebook
int sum = numbers.Sum();

Console.WriteLine($"Сумма всех элементов массива: {sum}");
```

```polyglot-notebook
double averageSum = numbers.Average();

Console.WriteLine($"Среднее арифметическое всех элементов массива: {averageSum}");
```

```polyglot-notebook
int[] sortedNumbers = numbers.OrderBy(x => x).ToArray();

Console.WriteLine("Отстортированный в порядке возрастания массив: " + string.Join(", ", sortedNumbers));
```

```polyglot-notebook
int num = 7;
bool contains = numbers.Contains(num);
Console.WriteLine($"Массив содержит число {num}: {contains}");
```
