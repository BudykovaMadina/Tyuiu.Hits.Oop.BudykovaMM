Создать базовый класс Book в C#, который будет представлять информацию о
книгах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

### Требования к базовому классу Book:
• Атрибуты: Название (Title), Автор (Author), Год издания (YearOfPublication).

• Методы:

    • GetInfo(): метод для получения информации о книге в виде строки.
    • Read(): метод для вывода сообщения о чтении книги.
    • Borrow(): метод для вывода сообщения о выдаче книги на чтение.

### Требования к производным классам:
1. Учебник (Textbook): Должен содержать дополнительные атрибуты, такие как
Предмет (Subject). Метод Read() должен быть переопределен для
добавления информации о предмете при чтении учебника.
2. Художественная литература (Fiction): Должен содержать дополнительные
атрибуты, такие как Жанр (Genre). Метод Borrow() должен быть
переопределен для добавления информации о жанре при выдаче книги на
чтение.
3. Научная литература (ScientificLiterature) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Область науки
(FieldOfScience). Метод GetInfo() должен быть переопределен для включения
информации об области науки в описании книги.


#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;

public interface ICopiable
{
    void MakeCopy();
    int CopyCount { get; set; }
}

public interface IAnnotatable
{
    void AddAnnotation(string note);
    string Annotation { get; set; }
}

public class Book : ICopiable
{
    public string Title { get; set; }
    public string Author { get; set; }
    public int YearOfPublication { get; set; }

    // +4 атрибута базового класса
    public int PageCount { get; set; }
    public decimal Price { get; set; }
    public bool IsAvailable { get; set; } = true;
    public int CopyCount { get; set; } = 1;

    public Book(string title, string author, int year, int pages, decimal price)
    {
        Title = title;
        Author = author;
        YearOfPublication = year;
        PageCount = pages;
        Price = price;
    }

    // +4 метода базового класса (с перегрузкой)
    public void Open() => Console.WriteLine($"Книга '{Title}' открыта.");
    public void Open(int page) => Console.WriteLine($"Книга '{Title}' открыта на странице {page}."); // перегрузка
    public void Close() => Console.WriteLine($"Книга '{Title}' закрыта.");
    public void UpdatePrice(decimal newPrice) { Price = newPrice; Console.WriteLine($"Цена '{Title}' обновлена до {Price:N2} руб."); }
    public void Archive() => Console.WriteLine($"Книга '{Title}' отправлена в архив.");

    public virtual void MakeCopy()
    {
        CopyCount++;
        Console.WriteLine($"Создана копия книги '{Title}'. Всего копий: {CopyCount}.");
    }

    public virtual string GetInfo() =>
        $"Книга: '{Title}', Автор: {Author}, Год: {YearOfPublication}, Страниц: {PageCount}, Цена: {Price:N2} руб.";

    // Перегрузка Read
    public virtual void Read() => Console.WriteLine($"Читаем книгу '{Title}'...");
    public virtual void Read(int pages) => Console.WriteLine($"Читаем '{Title}' — {pages} страниц.");
    public virtual void Read(string note) => Console.WriteLine($"Читаем '{Title}' ({note}).");

    public virtual void Borrow() => Console.WriteLine($"Книга '{Title}' выдана на чтение.");
}

// ==================== ПРОСТОЕ НАСЛЕДОВАНИЕ: Textbook ====================
public class Textbook : Book
{
    public string Subject { get; set; }

    // +4 атрибута
    public string EducationLevel { get; set; }
    public bool HasPracticalTasks { get; set; }
    public string ApprovalCode { get; set; }
    public int ExercisesCount { get; set; }

    public Textbook(string title, string author, int year, int pages, decimal price,
                    string subject, string level, bool practical,
                    string approval = "FGOS-2024", int exercises = 100)
        : base(title, author, year, pages, price)
    {
        Subject = subject;
        EducationLevel = level;
        HasPracticalTasks = practical;
        ApprovalCode = approval;
        ExercisesCount = exercises;
    }

    // +4 новых метода
    public void CheckCurriculum() => Console.WriteLine($"Проверка '{Title}' на соответствие ФГОС по предмету {Subject}.");
    public void SolveTest() => Console.WriteLine($"Решение упражнений из учебника '{Title}'.");
    public void AssignHomework(int chapter) => Console.WriteLine($"Задано ДЗ по главе {chapter} из '{Title}'.");
    public void VerifyApproval() => Console.WriteLine($"Код '{ApprovalCode}' для '{Title}' действителен.");

    // Перекрытие (override)
    public override string GetInfo() =>
        $"{base.GetInfo()}, Предмет: {Subject}, Уровень: {EducationLevel}";

    // Перекрытие Read + вызов базового через base
    public override void Read()
    {
        base.Read();
        Console.WriteLine($"  → учебник по '{Subject}' для {EducationLevel}.");
    }
}

// ==================== СЛОЖНОЕ НАСЛЕДОВАНИЕ: DigitalTextbook ====================
public class DigitalTextbook : Textbook
{
    public string FileFormat { get; set; }
    public double FileSizeMB { get; set; }

    // +4 атрибута
    public string CloudUrl { get; set; }
    public string EncryptionType { get; set; }
    public int DownloadCount { get; set; }
    public bool RequiresVpn { get; set; }

    public DigitalTextbook(string title, string author, int year, int pages, decimal price,
                           string subject, string level, bool practical,
                           string format, double size,
                           string url = "cloud.edu", string encryption = "AES-256")
        : base(title, author, year, pages, price, subject, level, practical)
    {
        FileFormat = format;
        FileSizeMB = size;
        CloudUrl = url;
        EncryptionType = encryption;
        RequiresVpn = true;
    }

    // +4 новых метода
    public void Download() => Console.WriteLine($"Загрузка '{Title}' в формате {FileFormat} ({FileSizeMB} MB)...");
    public void SyncWithCloud() => Console.WriteLine($"Синхронизация '{Title}' с {CloudUrl}...");
    public void Decrypt() => Console.WriteLine($"Расшифровка файла ({EncryptionType})...");
    public void IncrementDownloads() { DownloadCount++; Console.WriteLine($"Скачиваний '{Title}': {DownloadCount}."); }

    // Перекрытие
    public override string GetInfo() =>
        $"{base.GetInfo()}, Формат: {FileFormat}, Размер: {FileSizeMB} MB";

    // Перегрузка Download
    public void Download(bool highQuality) =>
        Console.WriteLine(highQuality
            ? $"Загрузка '{Title}' в HD-качестве ({FileSizeMB * 1.5} MB)..."
            : $"Загрузка '{Title}' в стандартном качестве ({FileSizeMB} MB)...");
}

// ==================== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: Fiction ====================
public class Fiction : Book, IAnnotatable
{
    public string Genre { get; set; }

    // +4 атрибута
    public string AgeRating { get; set; }
    public bool IsBestseller { get; set; }
    public string OriginalLanguage { get; set; }
    public int AwardsCount { get; set; }
    public string Annotation { get; set; } = "";

    public Fiction(string title, string author, int year, int pages, decimal price,
                   string genre, string ageRating, bool bestseller,
                   string language = "Русский", int awards = 0)
        : base(title, author, year, pages, price)
    {
        Genre = genre;
        AgeRating = ageRating;
        IsBestseller = bestseller;
        OriginalLanguage = language;
        AwardsCount = awards;
    }

    // +4 новых метода
    public void AnalyzePlot() => Console.WriteLine($"Анализ сюжета книги '{Title}'.");
    public void ReviewAgeRating() => Console.WriteLine($"Возрастной рейтинг '{Title}': {AgeRating}.");
    public void MarkBestseller() { IsBestseller = true; Console.WriteLine($"'{Title}' отмечена как бестселлер."); }
    public void TranslateTo(string lang) => Console.WriteLine($"'{Title}' переведена на {lang}.");

    // Перегрузка AddAnnotation
    public void AddAnnotation(string note)
    {
        Annotation = note;
        Console.WriteLine($"Аннотация к '{Title}': \"{note}\"");
    }
    public void AddAnnotation() => AddAnnotation("Без комментария."); // перегрузка

    // Перекрытие
    public override string GetInfo() =>
        $"{base.GetInfo()}, Жанр: {Genre}, Возраст: {AgeRating}, Бестселлер: {(IsBestseller ? "Да" : "Нет")}";

    public override void Borrow() =>
        Console.WriteLine($"Выдана художественная книга (Жанр: {Genre}, Возраст: {AgeRating}) — '{Title}'.");
}

// ==================== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: ScientificLiterature ====================
public class ScientificLiterature : Book, IAnnotatable
{
    public string FieldOfScience { get; set; }

    // +4 атрибута
    public string ResearchInstitution { get; set; }
    public bool PeerReviewed { get; set; }
    public string Doi { get; set; }
    public int ReferenceCount { get; set; }
    public string Annotation { get; set; } = "";

    public ScientificLiterature(string title, string author, int year, int pages, decimal price,
                                string field, string institution, bool peerReviewed,
                                string doi = "10.1000/xyz", int references = 50)
        : base(title, author, year, pages, price)
    {
        FieldOfScience = field;
        ResearchInstitution = institution;
        PeerReviewed = peerReviewed;
        Doi = doi;
        ReferenceCount = references;
    }

    // +4 новых метода
    public void VerifyCitations() => Console.WriteLine($"Проверка цитирований в '{Title}'.");
    public void PresentResearch() => Console.WriteLine($"Презентация исследования '{Title}' ({FieldOfScience}).");
    public void ShowDoi() => Console.WriteLine($"DOI: {Doi}.");
    public void ListReferences() => Console.WriteLine($"Источников: {ReferenceCount}.");

    // Перегрузка AddAnnotation
    public void AddAnnotation(string note)
    {
        Annotation = note;
        Console.WriteLine($"Научная рецензия к '{Title}': \"{note}\"");
    }
    public void AddAnnotation(int rating) => AddAnnotation($"Оценка рецензента: {rating}/10."); // перегрузка

    // Перекрытие
    public override string GetInfo() =>
        $"{base.GetInfo()}, Область: {FieldOfScience}, Институт: {ResearchInstitution}, Рецензия: {(PeerReviewed ? "Да" : "Нет")}";
}

// ==================== GENERIC КЛАСС: Catalog<T> ====================
public class Catalog<T> where T : Book
{
    private List<T> items = new List<T>();

    public void Add(T item) => items.Add(item);

    public void ShowAll()
    {
        Console.WriteLine($"--- Каталог ({typeof(T).Name}, {items.Count} шт.) ---");
        foreach (var item in items) Console.WriteLine(item.GetInfo());
    }

    public T FindByTitle(string title) => items.Find(b => b.Title == title);

    public int Count => items.Count;
}

// ==================== GENERIC КЛАСС: Repository<T> ====================
public class Repository<T> where T : Book
{
    private Dictionary<string, T> storage = new Dictionary<string, T>();

    public void Save(T item) => storage[item.Title] = item;

    public T Get(string title)
    {
        storage.TryGetValue(title, out var found);
        return found;
    }

    public void ShowAllTitles()
    {
        Console.WriteLine($"=== Хранилище ({typeof(T).Name}) ===");
        foreach (var key in storage.Keys) Console.WriteLine($" - {key}");
    }
}

// ==================== КЛАСС READER ====================
public class Reader
{
    public string Name { get; set; }
    private List<Book> borrowedBooks = new List<Book>();

    public Reader(string name) => Name = name;

    public void BorrowBook(Book book)
    {
        borrowedBooks.Add(book);
        Console.WriteLine($"{Name} взял книгу: '{book.Title}'");
        book.Borrow();
    }

    public void ReadBook(Book book)
    {
        Console.WriteLine($"{Name} читает...");
        book.Read();
    }

    // Перегрузка ReadBook с указанием количества страниц
    public void ReadBook(Book book, int pages)
    {
        Console.WriteLine($"{Name} читает {pages} стр. из книги...");
        book.Read(pages);
    }

    public void ShowBorrowed()
    {
        Console.WriteLine($"У {Name} на руках: {borrowedBooks.Count} книг.");
        foreach (var b in borrowedBooks) Console.WriteLine($"  - {b.Title}");
    }
}

// ==================== КЛАСС LIBRARY ====================
public class Library
{
    public string Name { get; set; }
    private List<Book> books = new List<Book>();

    public Library(string name) => Name = name;

    public void AddBook(Book book) => books.Add(book);

    public void LendBook(Book book, Reader reader)
    {
        if (!books.Contains(book))
        {
            Console.WriteLine($"Книги '{book.Title}' нет в библиотеке!");
            return;
        }
        books.Remove(book);
        Console.WriteLine($"Библиотека '{Name}' выдает '{book.Title}' читателю {reader.Name}");
        reader.BorrowBook(book);
    }

    // Перегрузка LendBook — без указания читателя (автоматически "Гость")
    public void LendBook(Book book) => LendBook(book, new Reader("Гость"));
}

// ==================== ДЕМОНСТРАЦИЯ ====================

var library = new Book[]
{
    new Textbook("Алгебра 9 класс", "А.Г. Мордкович", 2020, 250, 750.0m, "Математика", "Средняя школа", true),
    new DigitalTextbook("Физика 10 класс", "М.Я. Кандауров", 2023, 300, 450.0m, "Физика", "Старшая школа", true, "PDF", 15.4),
    new Fiction("Мастер и Маргарита", "М.А. Булгаков", 1967, 480, 990.0m, "Роман", "16+", true),
    new ScientificLiterature("Краткая история времени", "Стивен Хокинг", 1988, 220, 1200.0m, "Астрофизика", "Cambridge University", true)
};

Console.WriteLine("=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (перекрытие) ===");
foreach (var book in library)
{
    Console.WriteLine(book.GetInfo());     // override
    book.Read();                           // override
    book.Borrow();                         // override
    book.MakeCopy();                       // ICopiable
    book.Read(10);                         // перегрузка Read(int)
    book.Read("вдумчиво");                 // перегрузка Read(string)

    if (book is IAnnotatable annotatable)
    {
        annotatable.AddAnnotation("Обязательно к прочтению.");
    }
    Console.WriteLine("--------------------------------------------------");
}

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ ===");
var bookForTest = library[1];
bookForTest.Open();
bookForTest.Open(42);
((DigitalTextbook)bookForTest).Download();
((DigitalTextbook)bookForTest).Download(true);

Console.WriteLine("\n=== GENERIC КАТАЛОГ ===");
var catalog = new Catalog<Book>();
foreach (var b in library) catalog.Add(b);
catalog.ShowAll();
Console.WriteLine($"Найдено: {catalog.FindByTitle("Мастер и Маргарита")?.GetInfo()}");

var fictionCatalog = new Catalog<Fiction>();
fictionCatalog.Add((Fiction)library[2]);
fictionCatalog.ShowAll();

Console.WriteLine("\n=== GENERIC РЕПОЗИТОРИЙ ===");
var repo = new Repository<Book>();
foreach (var b in library) repo.Save(b);
repo.ShowAllTitles();
Console.WriteLine($"Из хранилища: {repo.Get("Краткая история времени").GetInfo()}");

Console.WriteLine("\n=== ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ ===");
var lib = new Library("Городская библиотека");
foreach (var b in library) lib.AddBook(b);

var reader = new Reader("Иван");
lib.LendBook(library[2], reader);
reader.ReadBook(library[2]);
reader.ReadBook(library[2], 50);       // перегрузка
reader.ShowBorrowed();
lib.LendBook(library[0]);              // перегрузка LendBook
```
